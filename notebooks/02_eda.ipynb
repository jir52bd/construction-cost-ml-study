```python
# Exploratory Data Analysis

## Objective
#The objective of this analysis is to explore the structure,
#distribution, and relationships among the variables in the
#construction cost dataset.

#The main target variable is:

#Total_Cost_crore_BDT 
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/raw/construction_cost_dataset_50_projects.csv")

df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Total_Cost_crore_BDT'], bins=20, kde=True)

plt.title('Distribution of Total Cost (crore BDT)')
plt.xlabel('Total Cost (crore BDT)')
plt.ylabel('Frequency')

plt.show()
```

```python
df['Total_Cost_crore_BDT'].describe()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x='Project_Duration_months',
    y='Total_Cost_crore_BDT'
)

plt.title('Total Cost vs Project Duration')
plt.xlabel('Project Duration (months)')
plt.ylabel('Total Cost (crore BDT)')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Number_of_Floors",
    y="Total_Cost_crore_BDT"
)

plt.title("Number of Floors vs Total Construction Cost")
plt.xlabel("Number of Floors")
plt.ylabel("Total Cost (Crore BDT)")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Material_Cost_crore_BDT",
    y="Total_Cost_crore_BDT"
)

plt.title("Material Cost vs Total Construction Cost")
plt.xlabel("Material Cost (Crore BDT)")
plt.ylabel("Total Cost (Crore BDT)")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Labour_Cost_crore_BDT",
    y="Total_Cost_crore_BDT"
)

plt.title("Labour Cost vs Total Construction Cost")
plt.xlabel("Labour Cost (Crore BDT)")
plt.ylabel("Total Cost (Crore BDT)")

plt.show()
```

```python
numeric_columns = [
    "Land_Area_sqft",
    "Total_Floor_Area_sqft",
    "Number_of_Floors",
    "Project_Duration_months",
    "Material_Cost_crore_BDT",
    "Labour_Cost_crore_BDT",
    "Total_Cost_crore_BDT"
]

correlation = df[numeric_columns].corr()

correlation
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix of Numerical Variables")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Building_Type",
    y="Total_Cost_crore_BDT"
)

plt.title("Construction Cost by Building Type")
plt.xlabel("Building Type")
plt.ylabel("Total Cost (Crore BDT)")

plt.show()
```

### Observation

The scatter plot indicates that total construction cost
generally increases as total floor area increases.

This suggests that total floor area may be an important
predictor of construction cost.
